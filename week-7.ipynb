**ACTIVITY 1**

````python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import numpy as np
````

````python
data = {
    'Size_cm': [7, 7.5, 6.8, 18, 19, 17.5, 8.5, 9, 8.2, 7.2],
    'Color_Intensity': [0.8, 0.75, 0.82, 0.6, 0.58, 0.62, 0.9, 0.88, 0.92, 0.78],
    'FruitType': ['Apple', 'Apple', 'Apple',
                  'Banana', 'Banana', 'Banana',
                  'Orange', 'Orange', 'Orange', 'Apple']
}

df = pd.DataFrame(data)
````

````python
X = df[['Size_cm', 'Color_Intensity']]
y = df[['FruitType']]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
````

````python
print("Fruit Classification Data (X_train head):\n", X_train.head())

print("\nFruit Type (y_train head):\n", y_train.head())
````

**ACTIVITY 1.2**

````python
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train.values.ravel())

y_pred = model.predict(X_test)

print("\nActual Fruit Types:")
print(y_test.values.ravel())

print("\nPredicted Fruit Types:")
print(y_pred)

accuracy = accuracy_score(y_test.values.ravel(), y_pred)

print("\nAccuracy:", accuracy)
print("Accuracy percentage:", accuracy * 100, "%")
````

**ACTIVITY 2**

````python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
````

````python
temp = np.array([10, 15, 20, 25, 30, 12, 18, 23, 28, 32]).reshape(-1, 1)

kwh = np.array([50, 60, 75, 90, 105, 55, 70, 85, 95, 110])
````

````python
X_train_elec, X_test_elec, y_train_elec, y_test_elec = train_test_split(
    temp, kwh, test_size=0.3, random_state=42
)
````

````python
print("\nElectricity Consumption Data (X_train_elec):\n",
      X_train_elec[:5])

print("\nElectricity Consumption (y_train_elec):\n",
      y_train_elec[:5])
````

**ACTIVITY 2.2**

````python
model = LinearRegression()
````

````python
model.fit(X_train_elec, y_train_elec)
````

````python
y_pred_elec = model.predict(X_test_elec)

print("Actual Electricity Consumption:")
print(y_test_elec)

print("\nPredicted Electricity Consumption:")
print(y_pred_elec)
````

````python
mse = mean_squared_error(y_test_elec, y_pred_elec)

print("Mean Squared Error:", mse)
````

**ACTIVITY 2.3**

````python
plt.scatter(X_train_elec, y_train_elec, label='Training Data')

plt.plot(X_train_elec, model.predict(X_train_elec),
         label='Regression Line')

plt.xlabel('Temperature (°C)')
plt.ylabel('Electricity Consumption (kWh)')
plt.title('Temperature vs Electricity Consumption')

plt.legend()
plt.show()
````
