```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
iris = load_iris(as_frame=True)
iris_df = iris.frame

print("Iris Data - First 5 Rows:")
print(iris_df.head())
```

```python
features = [
    'sepal length (cm)',
    'sepal width (cm)',
    'petal length (cm)',
    'petal width (cm)'
]

X = iris_df[features]

print("\nOriginal Features:")
print(X.head())
```

```python
minmax_scaler = MinMaxScaler()

X_normalized = minmax_scaler.fit_transform(X)

normalized_df = pd.DataFrame(
    X_normalized,
    columns=features
)

print("\nNormalized Data - First 5 Rows:")
print(normalized_df.head())

# Check range
print("\nMinimum values after normalization:")
print(normalized_df.min())

print("\nMaximum values after normalization:")
print(normalized_df.max())
```

```python
standard_scaler = StandardScaler()

X_standardized = standard_scaler.fit_transform(X)

standardized_df = pd.DataFrame(
    X_standardized,
    columns=features
)

print("\nStandardized Data - First 5 Rows:")
print(standardized_df.head())
```

```python
print("\nMean after standardization:")
print(standardized_df.mean())

print("\nStandard deviation after standardization:")
print(standardized_df.std())
```

```python
import pandas as pd

# Download dataset
!wget -q https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data

# Define column names
column_names = [
    'age',
    'workclass',
    'fnlwgt',
    'education',
    'education-num',
    'marital-status',
    'occupation',
    'relationship',
    'race',
    'sex',
    'capital-gain',
    'capital-loss',
    'hours-per-week',
    'native-country',
    'income'
]

# Load dataset
adult_df = pd.read_csv(
    'adult.data',
    header=None,
    names=column_names,
    skipinitialspace=True
)

# Display first 5 rows
print("Adult Income Data - First 5 Rows:")
print(adult_df.head())

# Check age range
print("\nMinimum age:", adult_df['age'].min())
print("Maximum age:", adult_df['age'].max())

# Define bins and labels
bins = [17, 30, 45, 60, adult_df['age'].max() + 1]

labels = [
    'Young Adult',
    'Middle-Aged',
    'Senior Adult',
    'Elderly'
]

# Discretize age
adult_df['age_group'] = pd.cut(
    adult_df['age'],
    bins=bins,
    labels=labels,
    right=False
)

# Display age and age group
print("\nAge and Age Group:")
print(adult_df[['age', 'age_group']].head(10))

# Display unique values and their counts
print("\nAge Group Counts:")
print(adult_df['age_group'].value_counts())
```

```python
import pandas as pd

# Step 1: Download MovieLens 100K
!wget -q https://files.grouplens.org/datasets/movielens/ml-100k.zip

# Step 2: Extract the ZIP file
!unzip -q ml-100k.zip

# Step 3: Load ratings data
ratings_df = pd.read_csv(
    'ml-100k/u.data',
    sep='\t',
    encoding='latin-1',
    header=None,
    names=['user_id', 'item_id', 'rating', 'timestamp']
)

print("Ratings Data:")
print(ratings_df.head())

# Step 4: Load movie information
movies_df = pd.read_csv(
    'ml-100k/u.item',
    sep='|',
    encoding='latin-1',
    header=None,
    usecols=[0, 1],
    names=['item_id', 'movie_title']
)

print("\nMovies Data:")
print(movies_df.head())

# Step 5: Calculate average rating and number of ratings
movie_aggregation = ratings_df.groupby('item_id').agg(
    avg_rating=('rating', 'mean'),
    num_ratings=('rating', 'count')
).reset_index()

print("\nAggregated Movie Data:")
print(movie_aggregation.head())

# Step 6: Merge aggregated data with movie titles
movie_stats_df = pd.merge(
    movies_df,
    movie_aggregation,
    on='item_id'
)

# Step 7: Keep only required columns
movie_stats_df = movie_stats_df[
    ['movie_title', 'avg_rating', 'num_ratings']
]

# Step 8: Sort by number of ratings
movie_stats_df = movie_stats_df.sort_values(
    by='num_ratings',
    ascending=False
)

# Step 9: Display top 10 movies
print("\nTop 10 Movies by Number of Ratings:")
print(movie_stats_df.head(10))
```
